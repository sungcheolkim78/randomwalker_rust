```rust
struct Walker {
    x: f64,
    y: f64,
    z: f64,
    t: f64,
    id: usize
}

impl Walker {
    fn next(&mut self) {
        self.x += 1.0;
        self.y += 1.0;
        self.z += 1.0;
        self.t += 1.0;
    }
    fn display(&self) {
        println!("Walker [{}] - ({}, {}, {}) - t={}", self.id, self.x, self.y, self.z, self.t);
    }
}
```

```rust
let mut w = Walker {
    x:0.0, y:0.0, z:0.0, t:0.0, id:1 
};
```

```rust
for i in 0..5 {
    w.next();
    w.display();
}
```
